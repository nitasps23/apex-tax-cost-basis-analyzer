```python
# Imports
import os
import pandas as pd
from sqlalchemy import create_engine
from dotenv import load_dotenv
```

```python
load_dotenv()

# Quick check to make sure it's working
print("DB_USER:", os.getenv('DB_USER'))
print("DB_NAME:", os.getenv('DB_NAME'))
# Don't print password - just confirm it loaded
print("Password loaded:", os.getenv('DB_PASSWORD') is not None)
```

```python
# Reads from data/ folder
engine = create_engine(
    f"postgresql://{os.getenv('DB_USER')}:{os.getenv('DB_PASSWORD')}@localhost/{os.getenv('DB_NAME')}"
)

project_root = os.path.dirname(os.path.abspath(''))
data_dir = os.path.join(project_root, 'data')

pd.read_csv(
    os.path.join(data_dir, 'transactions.csv')
).to_sql('transactions', engine, 
         if_exists='replace', index=False)

pd.read_csv(
    os.path.join(data_dir, 'cost_basis_lots.csv')
).to_sql('cost_basis_lots', engine, 
         if_exists='replace', index=False)

pd.read_csv(
    os.path.join(data_dir, 'tax_reporting.csv')
).to_sql('tax_reporting', engine, 
         if_exists='replace', index=False)

print("✓ All tables loaded successfully")
```
